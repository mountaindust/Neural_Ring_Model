# All-Unstable Self-Consistent Equilibria in NeuralBandModel

With power warping (`neural_angle='power'`) and delta function targets, the
NeuralBandModel finds spatial regions where **all** self-consistent equilibria
are unstable. This notebook visualizes the phenomenon via `plot_direction_mesh`.

In the direction mesh plot:
- **Blue** arrows = single stable equilibrium
- **Cyan** arrows = single **unstable** equilibrium
- **Black** arrows = one of multiple stable equilibria
- **Red** arrows = one of multiple **unstable** equilibria

Points that show only cyan or red arrows have no stable self-consistent
equilibrium — the animal cannot settle on a heading there.

```python
import numpy as np
import matplotlib.pyplot as plt
import decision_model as model
from multiprocessing import Pool

# Two delta-function targets in the standard geometry
target_locs = np.array([[4.33, 2.5], [4.33, -2.5]])
targets = model.Targets(locs=target_locs, geom_name=None)

# PerceptionModel with power warping (this is what triggers the bifurcation)
focal_loc = (0, 0)
focal_angle = 0
percep_model = model.PerceptionModel(targets, focal_loc, focal_angle,
                                     neural_weight=None, neural_angle='power')

# NeuralBandModel direction mesh — all-unstable regions will appear as
# cyan (single unstable) or all-red (multiple unstable) arrows
neur_model = model.NeuralBandModel(percep_model)

with Pool(10) as pool:
    neur_model.plot_direction_mesh(pool=pool, wb_plot=True)
```

## Comparison: same geometry without power warping

Without warping, every point has at least one stable equilibrium.

```python
# Same targets, no warping — for comparison
percep_model_nowarp = model.PerceptionModel(targets, focal_loc, focal_angle,
                                            neural_weight=None, neural_angle=None)
neur_model_nowarp = model.NeuralBandModel(percep_model_nowarp)

with Pool(10) as pool:
    neur_model_nowarp.plot_direction_mesh(pool=pool, wb_plot=True)
```
