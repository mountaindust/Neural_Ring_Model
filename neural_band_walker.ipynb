## Run random walks on neural band model with two targets

Use smooth cutoff function for both neural weight and warping

```python
import numpy as np
import matplotlib.pyplot as plt
import decision_model as model

target_locs = np.array([[4.33,2.5],[4.33,-2.5]]) # (x,y) coordinates

# targets = model.Targets(locs=target_locs, geom_name=None)
targets = model.Targets(locs=target_locs, geom_name='circle', r=0.5)

focal_loc = (0,0)
focal_angle = 0
percep_model = model.PerceptionModel(targets, focal_loc, focal_angle, 
                                     neural_weight=None, neural_angle=None)
percep_model_new = model.PerceptionModel(targets, focal_loc, focal_angle,
                                        neural_weight='cutoff', neural_angle='integral')
percep_model_new.a = np.pi/4
percep_model_new.b = np.pi*0.87

neur_model = model.NeuralBandModel(percep_model_new, K=1)
neur_model.rng = np.random.default_rng(seed = 3)
```

```python
neur_model.K =3 # Coupling strength for physical turning
neur_model.plot_walkers(dt=0.1, v=1, std=0.1, repetitions=5, max_steps=200,
                        start_loc=None, start_angle=None, #if None, uses what's set cell above
                        plot_tracks=True, wb_plot=True)
```
