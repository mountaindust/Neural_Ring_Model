## Compare Linear Cutoff function with Beta

```python
import numpy as np
import matplotlib.pyplot as plt
import decision_model as model

target_locs = np.array([[4.33,2.5],[4.33,-2.5]]) # (x,y) coordinates

# targets = model.Targets(locs=target_locs, geom_name=None)
targets = model.Targets(locs=target_locs, geom_name='circle', r=0.5)

focal_loc = (0,0)
focal_angle = 0
percep_model_sc = model.PerceptionModel(targets, focal_loc, focal_angle, 
                                        neural_angle_dist='lin_cutoff', 
                                        angle_weight='neural_angle_dist',
                                        a_warp=np.pi/4, b_warp=np.pi)
percep_model_beta = model.PerceptionModel(targets, focal_loc, focal_angle,
                                        neural_angle_dist='symmetric_beta', 
                                        angle_weight='neural_angle_dist',
                                        a_warp = 2.35, b_warp=np.pi) # alpha, b parameters

# Plot the graphs of the weighting functions for the two models
plt.figure(figsize=(12,6))
ax = plt.subplot()
percep_model_sc.plot_neural_weight(ax=ax)
percep_model_beta.plot_neural_weight(ax=ax)
plt.show()
```

```python
# Plot the perception model to see the current geometry and perception signal
percep_model_sc.plot_blocked_signals(wb_plot=True)
percep_model_beta.plot_blocked_signals(wb_plot=True)
```

```python
neur_model_sc = model.NeuralBandModel(percep_model_sc)
neur_model_beta = model.NeuralBandModel(percep_model_beta)

plt.figure(figsize=(12,6))
ax1 = plt.subplot(121)
ax2 = plt.subplot(122)

from multiprocessing import Pool
with Pool(10) as pool:
    ### ------ Directional mesh plots ------ ###
    # neur_model_sc.plot_direction_mesh(pool=pool, wb_plot=True, ax=ax1, title='Cutoff model equilibrium plot')
    # neur_model_beta.plot_direction_mesh(pool=pool, wb_plot=True, ax=ax2, title='Beta model equilibrium plot')
    ### ------ Bifurcation diagrams ------ ###
    neur_model_sc.plot_bifurcation_diagram(pool=pool, wb_plot=True, ax=ax1, 
                                           refinement_levels=4, max_count=3,
                                           num_x=41, num_y=41,
                                           title='Smooth cutoff model bifurcation diagram')
    neur_model_beta.plot_bifurcation_diagram(pool=pool, wb_plot=True, ax=ax2,
                                            refinement_levels=4, max_count=3,
                                            num_x=41, num_y=41,
                                            title='Beta model bifurcation diagram')
plt.legend(title='# stable\nequilibria')
plt.show()
```
