This notebook is to explore various aspects of continuum model for geometric-based decision-making.

```python
########## MUST RUN ##########

%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
import decision_model as model

# First, define the location and geometry of targets in Euclidean space as a Targets object.
#    Options and necessary parameters for the geometry of the targets is given in the constructor.
#    None: delta function geometry, no other parameters needed
#    'circle': targets are circles of radius r (they can share the same radius or have different ones)
#    'segment': targets are line segments with length l and orientation theta (shared or different for each)

target_locs = np.array([[20,5],[20,15]]) # (x,y) coordinates

targets = model.Targets(locs=target_locs, geom_name='circle', r=0.5)
```

```python
########## MUST RUN ##########

# Next, define an observer location and angle of observation in Euclidean space as a PerceptionModel object.
#    Pass in the Targets object to the constructor.

focal_loc = (0,10) #(20,5)
focal_angle = 1

percep_model = model.PerceptionModel(targets, focal_loc, focal_angle)

# You can plot the perception model to see the current geometry and perception signal
percep_model.plot(wb_plot=True)
```

```python
########## MUST RUN ##########

# Finally, define the weighting to use in the Hamiltonian and the type of consenus model within the context of a DirectionModel.
#    Pass in the PerceptionModel object to the constructor.
#    Options and necessary arguments for the weighting function can be found in the constructor documentation of DirectionModel.
#    Consensus type is either 'additive' or 'argmax'

weighting_name='truncnorm'
mu = 0
sigma = np.pi/8
left = -np.pi
right = np.pi

consensus_type='additive'

dir_model = model.DirectionModel(percep_model, consensus_type, weighting_name, mu, sigma, left, right)

# You can plot the weighting function to be sure it's what you want
# dir_model.plot_weighting(wb_plot=True)
```

```python
# We can plot the hamiltonian and signal for the current geometry. The black dot is the consensus direction.
dir_model.plot_hamiltonian(with_signal=True, wb_plot=True)
```

```python
# We can also loop over focal positions to observe how the Hamiltonian and perceieved signal vary by location

# Create a list of x,y coordinates
num_stps = 1000
focal_loc_mesh = np.column_stack((np.linspace(0,19,num_stps), 10*np.ones(num_stps)))

# Plot
dir_model.plot_hamiltonian(focal_loc_mesh, with_signal=True, wb_plot=True)
```

```python
# We can grid space to create a vector field of directions resulting from the model being evaluated at different
#    points around the targets.

dir_model.plot_direction_mesh(wb_plot=True)
```
