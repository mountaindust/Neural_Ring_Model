## Compare Smooth Cutoff function with Von Mises

```python
import numpy as np
import matplotlib.pyplot as plt
import decision_model as model

target_locs = np.array([[4.33,2.5],[4.33,-2.5]]) # (x,y) coordinates

# targets = model.Targets(locs=target_locs, geom_name=None)
targets = model.Targets(locs=target_locs, geom_name='circle', r=0.5)

focal_loc = (0,0)
focal_angle = 0
percep_model_sc = model.PerceptionModel(targets, focal_loc, focal_angle, 
                                     neural_weight='cutoff', neural_angle='integral')
percep_model_vm = model.PerceptionModel(targets, focal_loc, focal_angle,
                                        neural_weight='vonmises', neural_angle='integral')
percep_model_sc.a = np.pi/4
percep_model_sc.b = np.pi

percep_model_vm.k = 0.55

# Plot the graphs of the weighting functions for the two models
# plt.figure(figsize=(12,6))
# ax = plt.subplot()
# percep_model_sc.plot_neural_weight(ax=ax)
# percep_model_vm.plot_neural_weight(ax=ax)
# ax.legend(); plt.show()
```
